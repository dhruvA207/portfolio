```python
%%js

// CODE_RUNNER: Build a SpaceShip class like the one in Void Striker, then instantiate it and call its methods.

class SpaceShip {
  constructor(x, y, speed) {
    this.x = x;
    this.y = y;
    this.speed = speed;
    this.lives = 3;
  }
  move(dx, dy) {
    this.x += dx * this.speed;
    this.y += dy * this.speed;
  }
  status() {
    return `Ship at (${this.x.toFixed(1)}, ${this.y.toFixed(1)}) — lives: ${this.lives}`;
  }
}

const ship = new SpaceShip(512, 430, 4.5);
ship.move(1, 0);   // move right
ship.move(0, -1);  // move up
console.log(ship.status());
// Try: add a shoot() method, change the starting position, or swap in Shadow's speed (5.2)
```

```python
%%js

// CODE_RUNNER: Run the asteroid gravity simulation from Void Striker for 12 frames.

let verticalVelocity   =  0.5;   // positive = upward seed; asteroid starts slow then falls
const gravityAccel     =  0.055; // subtracted every frame — pulls the asteroid down
const terminalVelocity =  3.65;  // max fall speed; prevents infinite acceleration

for (let frame = 1; frame <= 12; frame++) {
  verticalVelocity -= gravityAccel;                    // step 1: gravity pulls down
  if (-verticalVelocity > terminalVelocity)            // step 2: clamp at terminal
    verticalVelocity = -terminalVelocity;
  const vy = -verticalVelocity;                        // step 3: flip sign for canvas y
  console.log(`Frame ${String(frame).padStart(2)}: fall speed = ${vy.toFixed(3)} px/frame`);
}
// Try: seed verticalVelocity = 2.0 — does it still respect the terminal velocity cap?
```

```python
%%js

// CODE_RUNNER: Simulate the WASD input system from Void Striker — no browser events needed.

const keys = {};
const ship  = { x: 512, y: 430, speed: 4.5 };
const W = 1025, H = 570;

function updateShip() {
  if (keys['a'] || keys['A']) ship.x -= ship.speed;
  if (keys['d'] || keys['D']) ship.x += ship.speed;
  if (keys['w'] || keys['W']) ship.y -= ship.speed;
  if (keys['s'] || keys['S']) ship.y += ship.speed;
  ship.x = Math.max(0, Math.min(W, ship.x));
  ship.y = Math.max(0, Math.min(H, ship.y));
}

// Simulate 5 frames holding D (right) + W (up)
keys['d'] = true; keys['w'] = true;
for (let f = 1; f <= 5; f++) {
  updateShip();
  console.log(`Frame ${f}: ship at (${ship.x.toFixed(1)}, ${ship.y.toFixed(1)})`);
}
// Try: change to keys['s'] + keys['a'] (down-left), or increase ship.speed to 10
```

```python
%%js

// CODE_RUNNER: Debug the chaser enemy AI from Void Striker — add logs to verify it closes distance each frame.

const ship  = { x: 400, y: 430 };
const enemy = { x: 200, y: 80, speed: 1.5, chaseAcc: 0 };

for (let frame = 1; frame <= 8; frame++) {
  const dx = ship.x - enemy.x;
  const dy = ship.y - enemy.y;
  const dist = Math.hypot(dx, dy) || 1;
  enemy.chaseAcc = Math.min(enemy.speed, enemy.chaseAcc + 0.0006);
  const spd = enemy.speed + enemy.chaseAcc;
  enemy.x += (dx / dist) * spd;
  enemy.y += (dy / dist) * spd;

  // What extra values would help you fully verify the pursuit math?
  console.log(`Frame ${frame}: pos=(${enemy.x.toFixed(1)}, ${enemy.y.toFixed(1)})  dist=${dist.toFixed(1)}  spd=${spd.toFixed(3)}`);
}
// Try: add Math.atan2(dy, dx) to confirm the angle points toward the ship
```

```python
%%js

// CODE_RUNNER: Run the Void Striker collision test suite — add your own edge-case tests at the bottom.

function circleCollide(ax, ay, ar, bx, by, br) {
  return Math.sqrt((ax - bx) ** 2 + (ay - by) ** 2) < ar + br;
}

const ship = { x: 500, y: 400, r: 20 };

const tests = [
  { label: 'Direct hit (enemy)',    ex: 510, ey: 410, er: 18, expect: true  },
  { label: 'Clean miss',            ex: 560, ey: 400, er: 18, expect: false },
  { label: 'Boss overlap',          ex: 500, ey: 360, er: 40, expect: true  },
  { label: 'Exact boundary (miss)', ex: 538, ey: 400, er: 18, expect: false },
  // Add your own: { label: '...', ex: ..., ey: ..., er: ..., expect: true/false },
];

let passed = 0;
tests.forEach(t => {
  const got = circleCollide(ship.x, ship.y, ship.r, t.ex, t.ey, t.er);
  const ok  = got === t.expect;
  if (ok) passed++;
  console.log(`${ok ? '\u2713' : '\u2717'} ${t.label}: ${ok ? 'PASS' : 'FAIL — got ' + got + ', expected ' + t.expect}`);
});
console.log('\n' + passed + '/' + tests.length + ' tests passed');
```
